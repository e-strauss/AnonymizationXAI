```python
from utils.explainer import FeatureImportanceGradient
import pandas as pd
```

```python
attack_df = pd.read_csv("../4_ExplanationResults/bracketed_words_with_clean_text.csv")
attack_df = attack_df.rename(columns={"Original Text (Cleaned)" : "text"})
```

```python
attack_df.columns
```

```python
explainer_gradient = FeatureImportanceGradient(max_length=256)
```

```python
short_attacks = (attack_df[attack_df["text"].apply(lambda x: len(explainer_gradient.tokenizer.encode(x))) <= 256]
                 .copy()
                 .drop(columns=["Attack Index"]))
```

```python
short_attacks["GradientWords"] = short_attacks["text"].apply(lambda x: explainer_gradient.get_feature_importance(x))
```

```python
short_attacks.to_csv("../5_Evaluation/AttackVsGradient.csv", index=False)
```

```python
for row in short_attacks.itertuples():
    print("Text: \n{}\n".format(row[2]))   
    print("Attack words: {}\n".format(row[1].lower()))
    print("Gradient words: [{}]\n".format(", ".join(row[3])))
    print("---------------------------------------------------------------------------------------------------------------")
```
